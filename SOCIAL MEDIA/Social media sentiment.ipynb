# Create datasets

```python
import pandas as pd
import random
from datetime import datetime, timedelta

# Options
platforms = ['Twitter', 'Instagram', 'Facebook']
regions = ['India', 'USA', 'UK', 'Canada', 'Australia']
topics = ['Product', 'Politics', 'Entertainment', 'Sports', 'Technology', 'Health', 'Education']
genders = ['Male', 'Female']

# Realistic Text Data
positive_texts = [
    "I recently bought this and it exceeded my expectations",
    "Absolutely loved the experience, highly recommended",
    "This is one of the best services I have ever used",
    "Amazing quality and fast delivery",
    "Very satisfied with the overall performance"
]

neutral_texts = [
    "The product is okay, nothing special about it",
    "Average experience, could be improved",
    "It works fine but has some minor issues",
    "Not too good, not too bad",
    "It is acceptable for the price"
]

negative_texts = [
    "Very disappointed with the product quality",
    "Waste of money, not recommended",
    "Had a terrible experience using this service",
    "Poor performance and bad customer support",
    "The product did not meet my expectations"
]

data = []

# Generate 5000 rows
for i in range(1, 5001):
    
    sentiment = random.choice(['Positive', 'Neutral', 'Negative'])
    
    if sentiment == 'Positive':
        text = random.choice(positive_texts)
        rating = random.randint(4, 5)
    elif sentiment == 'Neutral':
        text = random.choice(neutral_texts)
        rating = 3
    else:
        text = random.choice(negative_texts)
        rating = random.randint(1, 2)

    likes = random.randint(50, 1000)
    shares = random.randint(10, 300)
    comments = random.randint(5, 200)
    
    engagement = likes + shares + comments

    row = [
        i,
        datetime.now() - timedelta(days=random.randint(0, 365)),
        random.choice(platforms),
        random.choice(regions),
        random.choice(topics),
        random.randint(18, 60),
        random.choice(genders),
        text,
        likes,
        shares,
        comments,
        engagement,
        rating,
        sentiment
    ]

    data.append(row)

# Create DataFrame
df = pd.DataFrame(data, columns=[
    'Post_ID','DateTime','Platform','Region','Topic',
    'User_Age','User_Gender','Text','Likes','Shares',
    'Comments','Engagement','Rating','Sentiment'
])

# Save to Excel
df.to_excel("social_media_dataset.xlsx", index=False)

print("Dataset created successfully!")
df.head()
```
